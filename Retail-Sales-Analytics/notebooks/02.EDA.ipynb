```python
import pandas as pd
```

```python
df = pd.read_excel("../data/Professional_Retail_Sales_Dataset.xlsx")
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df["Month"] = df["Order Date"].dt.month_name()
df["Year"] = df["Order Date"].dt.year
df["Day"] = df["Order Date"].dt.day_name()
df["Profit Margin"] = (df["Profit"] / df["Sales"]) * 100
```

```python
df["Sales"].sum()
```

```python
print(f"Total Sales: ₹{df['Sales'].sum():,.2f}")
```

```python
print(f"Total Profit: ₹{df['Profit'].sum():,.2f}")
```

```python
print(f"Total Orders: {df['Order ID'].nunique()}")
```

```python
print(f"Total Customers: {df['Customer ID'].nunique()}")
```

```python
df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
```

```python
df.groupby("Category")["Profit"].sum().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt

category_sales = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(8,5))
category_sales.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
print(category_sales)
```

```python
category_profit = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)
print(category_profit)
```

```python
city_sales = df.groupby("City")["Sales"].sum().sort_values(ascending=False).head(10)
print(city_sales)
```

```python
product_sales = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)
print(product_sales)
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()
print(monthly_sales)
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()

plt.figure(figsize=(10,5))
monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)

plt.show()
```

```python
customer_name_sales = df.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False).head(5)
print(customer_name_sales)
```
